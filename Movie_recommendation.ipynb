```python
movies = movies[
    [
        "title",
        "genres",
        "keywords",
        "cast",
        "director",
        "overview",
        "tagline"
    ]
]
```

```python
movies.head()
```

```python
# selecting the relevant features for recommendation

selected_features = ['title','genres','keywords','overview','tagline','cast','director']
print(selected_features)

# replacing the null valuess with null string

for feature in selected_features:
  movies[feature] = movies[feature].fillna('')
```

```python
movies["tags"] = (
    movies["genres"] + " " +
    movies["keywords"] + " " +
    movies["cast"] + " " +
    movies["director"] + " " +
    movies["overview"]
)
```

```python
movies[["title", "tags", "genres"]].head()
```

```python
new_df = movies[["title", "tags", "genres"]]
```

```python
new_df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
vectors = cv.fit_transform(new_df["tags"])
```

```python
vectors.shape
```

```python
vectors.toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
similarity[0]
```

```python
movies_list = sorted(
    list(enumerate(similarity[0])),
    key=lambda x: x[1],
    reverse=True
)[1:6]
```

```python
print(movies_list[:10])
```

```python
def recommend(movie):

     movie_index = new_df[
     new_df["title"].str.lower() == movie.lower()
     ].index[0]

     distances = similarity[movie_index]

     movies_list = sorted(
        list(enumerate(distances)),
        key=lambda x: x[1],
        reverse=True
     )[1:6]

     recommendations = []

     for movie in movies_list:
        recommendations.append(
            new_df.iloc[movie[0]].title
        )

     return recommendations
```

```python
recommend("Avatar")
```
